##### Busca Dolar - Código VSCode - Awesomeapi

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
AWESOME_API_KEY = os.getenv("AWESOME_API_KEY")

#url = "https://economia.awesomeapi.com.br/json/last/USD-BRL"
url = "https://economia.awesomeapi.com.br/json/last/USD-BRL,USD-BRLT"
headers = {"x-api-key": AWESOME_API_KEY}
resposta = requests.get(url, headers=headers, timeout=10)
resposta.raise_for_status()
dolarc = resposta.json()["USDBRL"]
dolart = resposta.json()["USDBRLT"]

# print(dolarc)
print("Compra:", dolarc["bid"])
print("Venda:", dolarc["ask"])
print("Variação:", dolarc["pctChange"])
print("Data/hora:", dolarc["create_date"])
print("---------------------------------")
# print(dolart)
print("Compra:", dolart["bid"])
print("Venda:", dolart["ask"])
print("Variação:", dolart["pctChange"])
print("Data/hora:", dolart["create_date"])
```

#### Carga Hiostórica

```python
import os
import requests

AWESOME_API_KEY = os.getenv("AWESOME_API_KEY")

url = "https://economia.awesomeapi.com.br/json/daily/USD-BRL/30"
headers = {"x-api-key": AWESOME_API_KEY}

resposta = requests.get(url, headers=headers, timeout=30)

print("Status:", resposta.status_code)

if resposta.ok:
    dados = resposta.json()
    print("Registros:", len(dados))
    print("Primeiro:", dados[0] if dados else None)
    print("Último:", dados[-1] if dados else None)
else:
    print(resposta.text[:1500])
```
