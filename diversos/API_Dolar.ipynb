##### Busca Dolar - Código VSCode - Awesomeapi

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
AWESOME_API_KEY = os.getenv("AWESOME_API_KEY")

url = "https://economia.awesomeapi.com.br/json/last/USD-BRL"
headers = {"x-api-key": AWESOME_API_KEY}
resposta = requests.get(url, headers=headers, timeout=10)
resposta.raise_for_status()
dolar = resposta.json()["USDBRL"]

print(dolar)
print("Compra:", dolar["bid"])
print("Venda:", dolar["ask"])
print("Variação:", dolar["pctChange"])
print("Data/hora:", dolar["create_date"])
```
