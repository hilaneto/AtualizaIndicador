```python

import requests
from datetime import datetime, timezone, timedelta

url = "https://wis2bra.inmet.gov.br/oapi/collections/urn:wmo:md:br-inmet:synop/items"

agora = datetime.now(timezone.utc)
inicio = agora - timedelta(hours=2)
parametros = {"f": "json", "limit": 100, "datetime": f"{inicio.isoformat()}/{agora.isoformat()}"}

resposta = requests.get(url, params=parametros, timeout=30)
resposta.raise_for_status()
dados = resposta.json()
```

```python
temperaturas = []

for obs in dados["features"]:

    prop = obs["properties"]

    if prop["name"] == "air_temperature":

        temperatura = {
            "name": prop["name"],
            "value": prop["value"],
            "units": prop["units"],
            "phenomenonTime": prop["phenomenonTime"],
            "wigos_station_identifier":
                prop["wigos_station_identifier"]
        }

        temperaturas.append(temperatura)        
```

```python
capitais = [
    'BRASILIA', 'GOIANIA', 'CUIABA', 'CAMPO GRANDE',
    'MACEIO', 'SALVADOR', 'FORTALEZA', 'SAO LUIS',
    'JOAO PESSOA', 'RECIFE', 'TERESINA', 'NATAL',
    'ARACAJU', 'RIO BRANCO', 'MACAPA', 'MANAUS',
    'BELEM', 'PORTO VELHO', 'BOA VISTA', 'PALMAS',
    'VITORIA', 'BELO HORIZONTE', 'RIO DE JANEIRO',
    'SAO PAULO', 'CURITIBA', 'PORTO ALEGRE',
    'FLORIANOPOLIS'
]

# ---------------------------------------------------------
# 1. Localizar estações das capitais
# ---------------------------------------------------------
estacoes_capitais = {}
for estacao in estacoes:
    nome = estacao["nome"].upper()
    for capital in capitais:
        if capital in nome:
            estacoes_capitais[estacao["wigos"]] = {
                "capital": capital,
                "nome_estacao": estacao["nome"],
                "longitude": estacao["longitude"],
                "latitude": estacao["latitude"],
                "altitude": estacao["altitude"]
            }

for aa in estacoes_capitais:
    print(aa)
```

```python
# ---------------------------------------------------------
# 2. Localizar temperaturas dessas estações
# ---------------------------------------------------------

temperaturas = []

for obs in dados["features"]:
    prop = obs["properties"]
    wigos = prop["wigos_station_identifier"]
    if (prop["name"] == "air_temperature" and wigos in estacoes_capitais):
        estacao = estacoes_capitais[wigos]
        temperatura = {
            "capital": estacao["capital"],
            "estacao": estacao["nome_estacao"],
            "temperatura": prop["value"],
            "unidade": prop["units"],
            "dt_referencia": prop["phenomenonTime"],
            "wigos": wigos
        }

        temperaturas.append(temperatura)
```

```python
for temperatura in temperaturas:
    print(temperatura)
```
