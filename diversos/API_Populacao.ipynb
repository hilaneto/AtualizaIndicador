```python
import requests

url = "https://apisidra.ibge.gov.br/values/t/6579/n1/1/v/9324/p/all"

response = requests.get(url, timeout=30)
response.raise_for_status()

dados = response.json()[1:]

print("Registros:", len(dados))
print("Primeiro:", dados[0])
print("Último:", dados[-1])
```

```python
from datetime import date

registros = []

for item in dados:
    if item["V"].isdigit():
        registros.append({
            "populacao": int(item["V"]),
            "dt_referencia": date(int(item["D3C"]), 7, 1)
        })

registros

# print("Registros:", len(registros))
# print("Primeiro:", registros[0])
# print("Último:", registros[-1])
```
